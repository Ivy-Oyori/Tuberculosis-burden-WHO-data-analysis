```python
import pandas as pd
df = pd.read_csv ('TB_Burden_Country.csv')
print (df.shape)
```

```python
df.head()
```

```python
#dropping columns that are  empty 
#checking which ones are empty columns
empty_cols = df.columns[df.isnull().all()].tolist()
print("Fully empty columns:",empty_cols)

df = df.drop(columns = empty_cols)
```

```python
print(df.shape)
```

```python
#dropping other columns i might not need for analysis

cols_to_drop = [
    'ISO 2-character country/territory code',
    'Method to derive prevalence estimates',
    'Method to derive mortality estimates',
    'Method to derive incidence estimates',
    'Estimated prevalence of TB (all forms) per 100 000 population, low bound',
    'Estimated prevalence of TB (all forms) per 100 000 population, high bound',
    'Estimated prevalence of TB (all forms), low bound',
    'Estimated prevalence of TB (all forms), high bound',
    'Case detection rate (all forms), percent, low bound',
    'Case detection rate (all forms), percent, high bound',
]

df = df.drop(columns = [c for c in cols_to_drop if c in df.columns])
print(df.shape)
df.columns.tolist()
```

```python
#using regular expression in the naming of columnsand for pattern naming
import re

def to_snake_case(col):
    col = col.strip()
    col = col.replace('/', ' ').replace(',', '')
    col = re.sub(r'[^\w\s]', '', col)   # strip remaining punctuation
    col = re.sub(r'\s+', '_', col)       # spaces -> underscores
    return col.lower()

df.columns = [to_snake_case(c) for c in df.columns]
df.columns.tolist()
```

```python
#forgot to drop this column of ISO
df = df.drop(columns = ['iso_numeric_country_territory_code'])
df.shape
```

```python
#renaming the column names to short-form

df = df.rename(columns={
    'country_or_territory_name': 'country',
    'iso_3character_country_territory_code': 'iso3',

    'estimated_total_population_number': 'population',

    'estimated_prevalence_of_tb_all_forms_per_100_000_population': 'prevalence_rate',
    'estimated_prevalence_of_tb_all_forms': 'prevalence_cases',

    'estimated_mortality_of_tb_cases_all_forms_excluding_hiv_per_100_000_population': 'mortality_rate',
    'estimated_mortality_of_tb_cases_all_forms_excluding_hiv_per_100_000_population_low_bound': 'mortality_rate_low',
    'estimated_mortality_of_tb_cases_all_forms_excluding_hiv_per_100_000_population_high_bound': 'mortality_rate_high',

    'estimated_number_of_deaths_from_tb_all_forms_excluding_hiv': 'deaths',
    'estimated_number_of_deaths_from_tb_all_forms_excluding_hiv_low_bound': 'deaths_low',
    'estimated_number_of_deaths_from_tb_all_forms_excluding_hiv_high_bound': 'deaths_high',

    'estimated_mortality_of_tb_cases_who_are_hivpositive_per_100_000_population': 'mortality_rate_hiv',
    'estimated_mortality_of_tb_cases_who_are_hivpositive_per_100_000_population_low_bound': 'mortality_rate_hiv_low',
    'estimated_mortality_of_tb_cases_who_are_hivpositive_per_100_000_population_high_bound': 'mortality_rate_hiv_high',

    'estimated_number_of_deaths_from_tb_in_people_who_are_hivpositive': 'deaths_hiv',
    'estimated_number_of_deaths_from_tb_in_people_who_are_hivpositive_low_bound': 'deaths_hiv_low',
    'estimated_number_of_deaths_from_tb_in_people_who_are_hivpositive_high_bound': 'deaths_hiv_high',

    'estimated_incidence_all_forms_per_100_000_population': 'incidence_rate',
    'estimated_incidence_all_forms_per_100_000_population_low_bound': 'incidence_rate_low',
    'estimated_incidence_all_forms_per_100_000_population_high_bound': 'incidence_rate_high',

    'estimated_number_of_incident_cases_all_forms': 'incident_cases',
    'estimated_number_of_incident_cases_all_forms_low_bound': 'incident_cases_low',
    'estimated_number_of_incident_cases_all_forms_high_bound': 'incident_cases_high',

    'estimated_hiv_in_incident_tb_percent': 'hiv_in_incident_tb_pct',
    'estimated_hiv_in_incident_tb_percent_low_bound': 'hiv_in_incident_tb_pct_low',
    'estimated_hiv_in_incident_tb_percent_high_bound': 'hiv_in_incident_tb_pct_high',

    'estimated_incidence_of_tb_cases_who_are_hivpositive_per_100_000_population': 'incidence_rate_hiv',
    'estimated_incidence_of_tb_cases_who_are_hivpositive_per_100_000_population_low_bound': 'incidence_rate_hiv_low',
    'estimated_incidence_of_tb_cases_who_are_hivpositive_per_100_000_population_high_bound': 'incidence_rate_hiv_high',

    'estimated_incidence_of_tb_cases_who_are_hivpositive': 'incident_cases_hiv',
    'estimated_incidence_of_tb_cases_who_are_hivpositive_low_bound': 'incident_cases_hiv_low',
    'estimated_incidence_of_tb_cases_who_are_hivpositive_high_bound': 'incident_cases_hiv_high',

    'case_detection_rate_all_forms_percent': 'case_detection_rate',
})

print(df.shape)
df.columns.tolist()
```

```python
#checking data types
df.dtypes
```

```python
#checking for duplicates if any
#exact full-row duplicates
print("Exact duplicate rows:",df.duplicated().sum())

#also checking duplicate country and year pairs - they should be unique with one row per country per year
print("Duplicated country year pairs:", df.duplicated(subset =['country','year']).sum())
```

```python
#checking null patterns in HIV-related olumns to conifm the assumption of the early 2,000 ish
hiv_cols = [ c for c in df.columns if 'hiv' in c]
print(hiv_cols)

#also checking th e null rate by year for each HIV column
df.groupby('year')[hiv_cols].apply(lambda x: x.isnull().mean())
```

```python
# Which countries have NO hiv_in_incident_tb_pct estimate at all, across all years?
countries_missing_hiv_pct = df.groupby('country')['hiv_in_incident_tb_pct'].apply(lambda x: x.isnull().all())
print("Countries missing this estimate in ALL years:", countries_missing_hiv_pct.sum())

# Which countries have it in SOME years but not others?
countries_partial = df.groupby('country')['hiv_in_incident_tb_pct'].apply(lambda x: x.isnull().any() and not x.isnull().all())
print("Countries with partial missingness:", countries_partial.sum())
```

```python
#creating a new column to flag the the rows or countries with and without the hiv estimates with true and false
df['has_hiv_estimate'] = df['hiv_in_incident_tb_pct'].notnull()
df
```

```python
#lets flag the countries or years missing case detection rates as true or false
df['has_detection_estimate'] = df['case_detection_rate'].notnull()
df
```
